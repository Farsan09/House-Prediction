```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
train_df = pd.read_csv('Final_Train.csv')
test_df = pd.read_csv('Final_Test.csv')
```

```python
train_df.head()
```

```python
target = 'price'

X_train = train_df.drop(columns = [target])
y_train = train_df[target]

X_test = test_df.drop(columns = [target])
y_test = test_df[target]
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform (X_train)
X_test = scaler.transform (X_test)
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
lr_pred = lr.predict (X_test)
```

```python
print ('MAE: ', mean_absolute_error(lr_pred, y_test))
print ('\n')
print ('MSE: ', mean_squared_error(lr_pred, y_test))
print ('\n')
print ('RSS: ', r2_score(lr_pred, y_test))
```

```python
ridge = Ridge()
ridge.fit(X_train, y_train)
```

```python
ridge_pred = ridge.predict(X_test)
```

```python
print ('MAE: ', mean_absolute_error(ridge_pred, y_test))
print ('\n')
print ('MSE: ', mean_squared_error(ridge_pred, y_test))
print ('\n')
print ('RSS: ', r2_score(ridge_pred, y_test))
```

```python
lasso = Lasso(alpha = 1.0)
lasso.fit(X_train,y_train)
```

```python
lasso_pred = lasso.predict(X_test)
```

```python
print ('MAE: ', mean_absolute_error(lasso_pred, y_test))
print ('\n')
print ('MSE: ', mean_squared_error(lasso_pred, y_test))
print ('\n')
print ('RSS: ', r2_score(lasso_pred, y_test))
```

```python
elastic = ElasticNet(alpha=0.1, l1_ratio=0.5)
elastic.fit(X_train, y_train)
```

```python
elastic_pred = elastic.predict(X_test)
```

```python
print ('MAE: ', mean_absolute_error(elastic_pred, y_test))
print ('\n')
print ('MSE: ', mean_squared_error(elastic_pred, y_test))
print ('\n')
print ('RSS: ', r2_score(elastic_pred, y_test))
```
