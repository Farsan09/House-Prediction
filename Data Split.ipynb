```python
df['date'] = pd.to_datetime(df['date'])
print (df['date'].dt.year.value_counts())
```

### To avoid Leakage: The test set will run from 2012-2018
### The Eval Set will run from 2018-2022 and is unseen
### Holdout set will be from 2022 -2023

```python
df = df.sort_values('date')
```

```python
cut_off = '2020-01-01'
hold_out = '2022-01-01'
```

```python
### The training data should run from '12 to '18
train_df = df[df['date'] < cut_off]
```

```python
### The test data is between '18 and '22
test_df = df[(df['date'] > cut_off) & (df['date'] < hold_out)]
```

```python
### The holdout data is after '22
holdout_df = df[df['date'] > hold_out]
```

```python
###Saving to CSV
train_df.to_csv('Train Set.csv', index = False)
test_df.to_csv('Test Set.csv', index = False)
holdout_df.to_csv('HoldOut Set.csv', index = False)
```
